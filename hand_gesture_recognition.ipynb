```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset_path = r"C:\Users\divya\OneDrive\PRODIGY\PRODIGY_ML_04\rps-cv-images"

img_size = (150,150)
batch_size = 32
```

```python
train_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)
```

```python
val_ds = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=img_size,
    batch_size=batch_size
)
```
